# 4. Operadores relacionales

Los operadores relacionales permiten comparar valores. El resultado de una comparación es `TRUE` o `FALSE`.

- `==` : igual a
- `!=` : distinto de
- `<`  : menor que
- `>`  : mayor que
- `<=` : menor o igual que
- `>=` : mayor o igual que

Ejemplos:


```r
0 < 1
0 > 1
0 == 0
0 != 0
```

# 5. Operadores lógicos

Los operadores lógicos permiten combinar condiciones:

- `|` : OR; devuelve `TRUE` cuando al menos una condición es verdadera.
- `&` : AND; devuelve `TRUE` cuando ambas condiciones son verdaderas.
- `!` : NOT; invierte un valor lógico.


```r
TRUE | FALSE
TRUE & TRUE
TRUE & FALSE
!TRUE
```

## Ejemplo biomédico | ¿La muestra cumple los criterios de calidad?
