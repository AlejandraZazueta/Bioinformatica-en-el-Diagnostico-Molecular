## 3. Vector

Un vector reúne varios valores del mismo tipo.


````r
expresiones <- c(1.0, 1.2, 0.9, 2.1, 2.4, 1.9)
expresiones
````

````r
typeof(expresiones)
````

````r
class(expresiones)
````

````r
str(expresiones)
````

## 4. Matriz

Una matriz organiza datos del mismo tipo en filas y columnas.


````r
matriz_expresion <- matrix(
  expresiones,
  nrow = 2,
  ncol = 3,
  byrow = TRUE
)

matriz_expresion
````

````r
str(matriz_expresion)
````

## 5. Factor

Un factor representa una variable categórica con niveles definidos.


````r
grupo <- factor(c(
  rep("Control", 3),
  rep("Tratamiento", 3)
))

grupo
````

````r
levels(grupo)
````

````r
str(grupo)
````

## 6. Data frame

Un `data.frame` permite combinar columnas de diferentes tipos.


````r
datos_demo <- data.frame(
  muestra = paste0("M0", 1:6),
  grupo = grupo,
  expresion = expresiones,
  detectado = c(TRUE, TRUE, TRUE, TRUE, TRUE, TRUE)
)

datos_demo
````

````r
str(datos_demo)
````

## 7. Lista

Una lista puede reunir objetos de diferentes tipos y estructuras.


````r
resultado <- list(
  gen = "BRCA1",
  expresion = expresiones,
  resumen = datos_demo
)

resultado
````

````r
resultado$gen
````

````r
str(resultado)
````
