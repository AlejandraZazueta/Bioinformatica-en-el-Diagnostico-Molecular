## 4. ¿Cómo se construye un gráfico en `ggplot2`?

Antes de aprender distintos tipos de gráficos, observaremos cómo se construye **una figura capa por capa**.

La lógica general será:

**DATOS → ESTÉTICAS → GEOMETRÍA → CAPAS → ETIQUETAS → TEMA**

Trabajaremos con esta pregunta:

> **¿Cómo se distribuye el biomarcador entre los grupos experimentales?**

Tenemos:

- `grupo` → variable categórica
- `biomarcador` → variable numérica

Por ello, utilizaremos un **boxplot**.


### Capa 1 | Datos y estéticas

Primero indicamos:

- qué conjunto de datos utilizaremos;
- qué variable irá en el eje X;
- qué variable irá en el eje Y.


````r
p <- ggplot(
  datos,
  aes(x = grupo, y = biomarcador)
)

p
````

### Observa

Todavía no vemos los datos representados.

`ggplot()` creó la estructura del gráfico, pero aún no hemos indicado **cómo queremos representar las observaciones**.

Para eso necesitamos una geometría.


### Capa 2 | Geometría


````r
p <- p +
  geom_boxplot()

p
````

Ahora aparece la distribución de `biomarcador` para cada `grupo`.

`geom_boxplot()` indica **cómo se representarán visualmente los datos**.

En `ggplot2`, diferentes preguntas pueden requerir diferentes geometrías.


### Capa 3 | Agregar las observaciones individuales

Podemos agregar otra capa sin eliminar el boxplot.


````r
p <- p +
  geom_jitter(
    width = 0.10,
    size = 2,
    alpha = 0.70
  )

p
````

El boxplot sigue presente y ahora se agregaron los puntos.

Esto muestra una idea central de `ggplot2`:

> **Las capas se van sumando mediante `+`.**


### Capa 4 | Etiquetas

Ahora mejoraremos la comunicación de la figura.


````r
p <- p +
  labs(
    title = "Biomarcador según grupo experimental",
    x = "Grupo experimental",
    y = "Biomarcador (pg/mL)"
  )

p
````

Los datos no cambiaron. Solo hicimos la figura más clara para quien la interpreta.


### Capa 5 | Incorporar color

Podemos utilizar una variable para controlar una característica visual.


````r
p <- p +
  aes(fill = grupo)

p
````

### Dentro o fuera de `aes()`

Esta diferencia es importante:

```r
aes(fill = grupo)
```

El color representa una **variable**.

En cambio:

```r
geom_boxplot(color = "black")
```

establece una característica **fija**.

Por eso debemos preguntarnos siempre:

> ¿Quiero que esta característica represente información de mis datos o solo quiero modificar la apariencia?


### Capa 6 | Tema


````r
p <- p +
  theme_classic(base_size = 13) +
  theme(legend.position = "none")

p
````

## La figura se construyó paso a paso

Partimos con:

**datos + variables**

y fuimos agregando:

**geometría + observaciones + etiquetas + color + tema**

La estructura general que veremos repetidamente será:

```r
ggplot(datos, aes(...)) +
  geom_...() +
  labs(...) +
  theme_...()
```

Ahora podemos estudiar diferentes geometrías sin tener que aprender `ggplot2` nuevamente desde cero.

## 4. Histograma: distribución de una variable numérica

**Pregunta:** ¿cómo se distribuyen las concentraciones del biomarcador?

````r
ggplot(datos, aes(x = biomarcador)) +
  geom_histogram(binwidth = 5, fill = "#2C7FB8", color = "white") +
  labs(
    title = "Distribución del biomarcador",
    x = "Concentración (pg/mL)",
    y = "Número de muestras"
  ) +
  theme_minimal(base_size = 13)
````

### Demostración

Cambia `binwidth = 5` por `binwidth = 2` y luego por `binwidth = 12`.

- ¿La apariencia de la distribución cambia?
- ¿Cambió el conjunto de datos o solamente la manera de representarlo?

## 5. Boxplot: comparar una variable numérica entre grupos

**Pregunta:** ¿cómo varía el biomarcador entre los tres grupos experimentales?

````r
colores_grupos <- c(
  "Control" = "#6BAED6",
  "Tratamiento A" = "#74C476",
  "Tratamiento B" = "#FD8D3C"
)

ggplot(datos, aes(x = grupo, y = biomarcador, fill = grupo)) +
  geom_boxplot(width = 0.65, alpha = 0.85, outlier.shape = NA) +
  geom_jitter(width = 0.10, size = 2, alpha = 0.65) +
  scale_fill_manual(values = colores_grupos) +
  labs(
    title = "Biomarcador según grupo experimental",
    x = NULL,
    y = "Concentración (pg/mL)"
  ) +
  theme_classic(base_size = 13) +
  theme(legend.position = "none")
````

### Lectura responsable del boxplot

El gráfico permite describir mediana, dispersión y posibles valores extremos. Por sí solo **no demuestra significancia estadística** ni causalidad. Primero describimos lo observado y luego, si corresponde, aplicamos una prueba estadística adecuada.

## 6. Gráfico de barras: frecuencias de una variable categórica

**Pregunta:** ¿cuántas muestras respondedoras y no respondedoras hay en cada grupo?

````r
ggplot(datos, aes(x = grupo, fill = respuesta)) +
  geom_bar(position = "dodge", color = "white") +
  scale_fill_manual(values = c("No respondedor" = "#BDBDBD",
                               "Respondedor" = "#756BB1")) +
  labs(
    title = "Respuesta observada por grupo",
    x = NULL,
    y = "Número de muestras",
    fill = "Respuesta"
  ) +
  theme_minimal(base_size = 13)
````

### Idea clave

`geom_bar()` cuenta observaciones automáticamente. Un gráfico de barras no debe utilizarse para ocultar la distribución de una variable numérica. Si la pregunta compara concentraciones entre grupos, el boxplot anterior es más informativo.

## 7. Scatter plot: relación entre dos variables numéricas

**Pregunta:** ¿existe una relación entre la concentración del biomarcador y la expresión relativa del gen?

````r
ggplot(datos, aes(x = biomarcador, y = expresion_gen, color = grupo)) +
  geom_point(size = 2.8, alpha = 0.80) +
  geom_smooth(method = "lm", se = TRUE, color = "#333333") +
  scale_color_manual(values = colores_grupos) +
  labs(
    title = "Relación entre biomarcador y expresión génica",
    subtitle = "La línea representa un ajuste lineal con su intervalo de confianza",
    x = "Biomarcador (pg/mL)",
    y = "Expresión relativa",
    color = "Grupo"
  ) +
  theme_classic(base_size = 13)
````

### Interpretación cuidadosa

Una tendencia ascendente sugiere una asociación positiva, pero el gráfico no permite concluir causalidad. También debemos comprobar si la relación es aproximadamente lineal, si existen puntos influyentes y si el patrón cambia entre grupos.
