# 1. Cargar y conocer los datos

Coloque `datos_salud.csv` en la misma carpeta que este archivo.



```r
datos <- read.csv("datos_salud.csv")
head(datos)
str(datos)
summary(datos)
```


En este análisis utilizaremos:

- **X = `imc`** → variable predictora
- **Y = `presion_sistolica`** → variable respuesta

> Antes de construir un modelo, debemos saber qué variables tenemos y cuál será el rol de cada una.

# 2. Antes de modelar: observar la relación



```r
library(ggplot2)

ggplot(datos, aes(x = imc, y = presion_sistolica)) +
  geom_point() +
  labs(
    x = "IMC (kg/m²)",
    y = "Presión arterial sistólica (mmHg)",
    title = "IMC y presión arterial sistólica"
  ) +
  theme_minimal()
```


## Observemos juntos

Antes de calcular ningún estadístico, fíjese en:

- **Dirección:** ¿positiva o negativa?
- **Forma:** ¿aproximadamente lineal?
- **Dispersión:** ¿qué tan separados están los puntos?
- **Observaciones inusuales:** ¿hay puntos que llamen la atención?

> **Primero observamos la relación; después decidimos cómo modelarla.**

# 3. Cuantificar la asociación: correlación



```r
cor(datos$imc, datos$presion_sistolica)
```


El coeficiente `r` resume la **dirección** y la **magnitud** de la asociación lineal.

En estos datos obtendremos aproximadamente:

**r ≈ 0.63**

Por lo tanto, existe una **asociación lineal positiva** entre IMC y presión arterial sistólica.

### Pausa breve

**¿El valor de `r` demuestra que un mayor IMC causa una mayor presión arterial?**

> **No.** La correlación describe asociación, no causalidad.

# 4. Construir nuestro primer modelo lineal



```r
modelo <- lm(
  presion_sistolica ~ imc,
  data = datos
)
```


La instrucción:

`presion_sistolica ~ imc`

significa:

- **Y = `presion_sistolica`**
- **X = `imc`**

R buscará la recta:

**Ŷ = β̂₀ + β̂₁X**

que minimiza la suma de los residuos al cuadrado.

# 5. Examinar el modelo con `summary()`



```r
summary(modelo)
```


Buscaremos cuatro elementos:

1. **Estimate**
2. **Std. Error**
3. **Pr(>|t|)**
4. **Multiple R-squared**

# 6. ¿Qué aprendió el modelo?



```r
coef(modelo)
```


Con estos datos obtendremos aproximadamente:

- **Intercepto:** `β̂₀ ≈ 86.78`
- **Pendiente:** `β̂₁ ≈ 0.95`

Por lo tanto:

**Ŷ = 86.78 + 0.95 × IMC**

## Interpretación de la pendiente

> Por cada aumento de **1 kg/m² en IMC**, el modelo predice, en promedio, un aumento aproximado de **0.95 mmHg** en la presión arterial sistólica.

## Interpretación del intercepto

> El intercepto representa la presión arterial sistólica predicha cuando `IMC = 0`.

En este contexto, un IMC de 0 kg/m² no es biológicamente plausible, por lo que el intercepto es necesario para definir la recta, pero **no tiene una interpretación práctica útil**.

# 7. Incertidumbre y evidencia estadística



```r
summary(modelo)
```


En la fila correspondiente a `imc` encontramos:

- `Estimate` → cuánto estimamos;
- `Std. Error` → incertidumbre de la estimación;
- `Pr(>|t|)` → evidencia estadística respecto de una pendiente igual a 0.

En estos datos, el p-value para `imc` es muy pequeño (`p < 0.001`).

> Existe evidencia estadística de que la pendiente es distinta de cero, compatible con una asociación lineal entre IMC y presión arterial sistólica.

**Importante:** un p-value pequeño **no demuestra causalidad**.

# 8. ¿Cuánto explica el modelo?

En `summary(modelo)` busque `Multiple R-squared`.

En estos datos:

**R² ≈ 0.396**

Esto significa:

> El modelo lineal con IMC explica aproximadamente el **39.6 % de la variabilidad observada en la presión arterial sistólica**.

`R²` **no significa** porcentaje de exactitud ni probabilidad de que el modelo sea correcto.

# 9. Visualizar la recta que encontró el modelo



```r
ggplot(datos, aes(x = imc, y = presion_sistolica)) +
  geom_point() +
  geom_smooth(method = "lm", se = FALSE) +
  labs(
    x = "IMC (kg/m²)",
    y = "Presión arterial sistólica (mmHg)",
    title = "Regresión lineal: IMC y presión arterial sistólica"
  ) +
  theme_minimal()
```


Los puntos son las observaciones reales.

La recta representa las **predicciones del modelo**.

La distancia vertical entre cada punto y la recta corresponde al **residuo**.

# 10. Utilizar el modelo para predecir

Suponga un nuevo paciente con:

**IMC = 30 kg/m²**



```r
nuevo_paciente <- data.frame(imc = 30)

predict(modelo, newdata = nuevo_paciente)
```


La predicción será aproximadamente:

**Ŷ ≈ 115.3 mmHg**

El modelo predice una presión arterial sistólica aproximada de 115.3 mmHg para una persona con IMC = 30 kg/m².

### Pregunta clave

**¿Significa que una persona con IMC = 30 necesariamente tendrá exactamente 115.3 mmHg?**

> **No.** Es una predicción del modelo. Existe variabilidad que el modelo no explica.
